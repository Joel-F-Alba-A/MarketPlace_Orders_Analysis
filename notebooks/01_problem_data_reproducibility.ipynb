## **1. Problem and Question of Interest**

Commercial datasets usually combine information about customers, products, sales, profits, and shipping. However, their size and the hidden relationships between variables make it difficult to identify transaction profiles, investigate latent factors underlying commercial behavior, and recognize unusual transactions using traditional statistical summaries alone. Transactions can exhibit different profit margins, while atypical transactions may have either favorable or unfavorable business implications.

This study seeks to discover and characterize these structures without predefined labels. Different unsupervised learning algorithms are used to identify probabilistic transaction profiles, extract latent factors, visualize nonlinear relationships, investigate customer–product affinities and associations, and detect potentially anomalous orders.

**Question of Interest**

What structures, relationships, and latent patterns can be identified in commercial and operational order data through different unsupervised learning methods, and to what extent can these methods characterize transaction profiles, represent relationships between variables, explore customer–product affinities, and identify atypical orders?

The study covers GMM; SVD, PCA, and PPCA; FA and ICA; matrix-factorization recommendations; t-SNE and UMAP; statistical and density/tree-based anomaly detection; AE and VAE; and product association rules. Anomaly detection is exploratory: verified anomaly labels are unavailable.

## **4. Preprocessing Decisions**

The cleaning is implemented in notebook 00 using reusable functions from `src/preprocessing/`. It preserves every observation and exports the cleaned dataset in its original measurement units.

| Decision | Variables and justification |
|---|---|
| Remove highly incomplete fields | `Product Description` is entirely missing; `Order Zipcode` has 86.24% missing values. |
| Remove constant fields | `Customer Email`, `Customer Password`, and `Product Status` do not vary in the observed dataset. |
| Remove redundant information | `Order Item Product Price`, `Order Profit Per Order`, `Order Item Total`, `Product Category Id`, `Order Item Cardprod Id`, and `Order Customer Id` duplicate information available in retained fields. |
| Remove unused identifiers | Drop `Order Id` and `Order Item Id` from the modeling dataset; retain customer and product IDs for interaction matrices. Drop `Department Id` and `Category Id` while retaining their descriptive names. |
| Remove personal or unused fields | Drop `Customer Fname`, `Customer Lname`, `Customer Street`, and `Product Image`, which are outside the analytical scope. |
| Summarize dates | Extract `Order Month` from `order date (DateOrders)` and remove both original order and shipping timestamps. Month does not preserve year or within-month chronology. |
| Preserve meaningful extremes | Retain negative benefits and profit ratios because losses are not automatically data errors. No blanket outlier removal, logarithmic transformation, or winsorization is applied. |
| Preserve remaining missing values | Retain the three missing `Customer Zipcode` values; that variable is not part of the numerical model feature set. No row deletion or missing-value imputation is performed. |

In total, **21 original columns are removed and one derived column is added**. Removing order IDs limits subsequent order-level basket analysis; association rules in notebook 08 therefore describe customer purchase histories.

## **7. Scaling and Model-Specific Transformations**

Shipping days, monetary amounts, quantities, and proportions differ in units and range. Large-range variables can otherwise dominate distances, variance-based projections, density estimates, and neural reconstruction errors. Where appropriate, standardization uses:

$$z_{ij}=\frac{x_{ij}-\mu_j}{\sigma_j}.$$

| Notebook | Transformation and fitting scope |
|---|---|
| 02 — GMM | StandardScaler fitted on the 80% training partition; reused for validation. The GMM feature subset is Sales, Order Item Discount Rate, Order Item Profit Ratio, and Order Item Quantity. |
| 03 — SVD/PCA/PPCA | 70%/15%/15% partitions. PCA and PPCA use training-fitted centering/scaling. SVD uses `StandardScaler(with_mean=False)` to scale without centering. |
| 04 — FA/ICA and recommendation | FA/ICA use training-fitted standardization with 70%/15%/15% partitions. Recommendation aggregates quantities by customer and product into a sparse interaction matrix. |
| 05 — t-SNE/UMAP | Standardization fitted on the full numerical dataset before drawing a 30,000-row stratified sample. This is exploratory geometry analysis, not held-out predictive validation. |
| 06 — Anomaly detection | 60%/20%/20% partitions; scaler and score thresholds fitted using training data. Isolation Forest shares this input pipeline although it does not inherently require standardization. |
| 07 — AE/VAE | 60%/20%/20% partitions with training-fitted scaling. Reconstruction errors use standardized units; descriptive interpretation uses original values. |
| 08 — Association rules | Aggregate purchased quantities by customer and product, then convert to Boolean purchase presence. No standardization is required for support, confidence, and lift. |

Training-only fitting prevents validation/test statistics from determining the transformation in held-out analyses. Standardization does not make distributions Gaussian, eliminate skewness, or remove unusual observations. The processed CSV itself remains unscaled. Correlated and algebraically related features remain, including sales approximately equal to product price times quantity.

## **9. Seeds and Controlled Stability Experiments**

The principal seed is **42**, defined as `SEED` in `src/config.py`. It is used for reference partitions, sampling, and stochastic estimators where appropriate. Recommendation holdouts use seeds 42 and 43.

Additional seeds are used under controlled experimental conditions to evaluate sensitivity rather than to select a favorable run. When studying initialization sensitivity, partitions, scaling, features, and hyperparameters remain fixed while the model seed changes. This distinguishes initialization effects from changes in the data.

| Experiment | Seeds and interpretation |
|---|---|
| GMM | `[0, 10, 21, 42, 100, 123, 202, 500, 1000, 2026]`; repeated initializations and comparisons of initialization settings. |
| t-SNE, UMAP, Isolation Forest, AE, VAE | `[0, 21, 42, 123, 7, 13, 57, 99, 202, 314]`; repeated model runs in the relevant stability experiments. |
| LOF | The same ten-seed list controls 80% training-subset selection. LOF has no random initialization; the experiment measures sensitivity to reference samples. |
| PCA/SVD/PPCA and FA/ICA | Controlled bootstrap experiments use reference random generators and derived estimator seeds where implemented. These also assess sensitivity to resampling, rather than only initialization on identical observations. |
| Cleaning and association matrix | Column removal, month extraction, and customer–product aggregation are deterministic and require no random seed. |

Neural training routines control Python, NumPy, and PyTorch generators. Saved checkpoints, experiment metadata, and input fingerprints help trace cached results. Fixing seeds facilitates repetition but does not guarantee identical floating-point results across dependency versions, CPU/GPU hardware, or parallel numerical routines.
